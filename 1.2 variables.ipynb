Variables

Variable are fundemental elements in programming used to store data that can be refernced and manipulated in a program. In python,variables are created when assign a value to them,and they do not need explicit declaration to reserve memory space.The declaration happens automatically when you assign a value to a variable.

```python
## Declaring and Assigning variable

age=32
height=6.1
name="jyothi"
is_student=True

print("age:",age)
print("height:",height)
print("name:",name)
print("valid/Not:",is_student)
```

```python
## Naming conventions 
## Variable names should be descriptive
## They must start with a letter or an '_' and contains letters,numbers and undersacores
## variables names case sensitive

#valid variable name

first_name="jyothi"
last_name="prakash"
```

```python
##invalid variable names

# 2age=30
# first-name="jyothi"
# @name="prakash"
```

```python
#case sensitivity
name="jyothi"
Name="prakash"

name==Name
```

```python
## Understanding the variable types
## python is dynamically types,type of a variable is determined at runtime

age=25 #int
height=6.1 # float
name="krish"#string
is_student=True # bool

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

```python
# type checking and conversion

age=25
print(age)
print(type(age))

age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
## Dynamic typing 
## python allows the type of a variable to changes as the program executes

var=10 # int
print(var,type(var))

var="hello"
print(var,type(var))

var=3.15
print(var,type(var))
```
