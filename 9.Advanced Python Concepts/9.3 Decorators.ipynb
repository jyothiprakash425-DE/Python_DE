Decorators:

Decorators are a powerful and flexible feature in python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators,including how to create and use them.

```python
## function copy
## closures 
## decorators
```

```python
## function copy

def welcome():
    return "Welcome to that advanced python course"

welcome()
```

```python
wel=welcome
print(wel())
del welcome
print(wel())
```

```python
## closure functions

def main_welcome(msg):

    def sub_welcome_method():
        print("Welcome to that advanced python course")
        print(msg)
        print("Please learn these concepts properly")
        
    return sub_welcome_method()
```

```python
main_welcome("welcome everyone")
```

```python
def main_welcome(func):

    def sub_welcome_method():
        print("Welcome to that advanced python course")
        func("Welcome everyone tho this tutorial.")
        print("Please learn these concepts properly")
        
    return sub_welcome_method()
```

```python
main_welcome(print)
```

```python
def main_welcome(func,lst):

    def sub_welcome_method():
        print("Welcome to that advanced python course.")
        print(func(lst))
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(len,[1,2,3,4,5,6])
```

```python
## Decorator 

def main_welcome(func):

    def sub_welcome_method():
        print("Welcome to that advanced python course.")
        func()
        print("Please learn these concepts properly")
    
    return sub_welcome_method()
```

```python
@main_welcome
def course_introduction():
    print("This is an advanced python course.")

```

```python
def my_decorator(func):

    def wrapper():
        print("Somethins is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("hello")
```

```python
say_hello()
```

```python
## Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Hello")
```

```python
say_hello()
```
