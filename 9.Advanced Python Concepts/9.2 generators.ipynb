Generators:

Generators are simpler way to create iterators.They use the yield keyword to produce a series of values lazily.which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(n):
        return i**2
```

```python
square(3)
```

```python
def square(n):
    for i in range(n):
        yield i**2
```

```python
square(3)
```

```python
for i in range(3):
    print(i)
```

```python
a=square(3)
a
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen=my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

Practicle Example: reading Large FIles

generators are particularly used for reading large files because they allow you to process one line at a time without loading the entire file into memory.

```python
## reading large files

def reading_large_files(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_path="large_file.txt"

for line in reading_large_files(file_path):
    print(line)

```

Conclusion :

Iterators and generators are powerful tools in python for creating and handling sequences of data efficiently.Iterators provide a way to access elements sequentially,while generators allow you to generate items on the fly,making them particularly useful for handling large datasets and infinite sequences.Understanding these concepts will enable you to write more efficient and memory-concious python programs.
