Inheritance in Python:

Inheritance is a fundemental concept in Object-oriented programming(OOP) thats allows a class to inherit attributes and methods from another class.This lession cobers single inheritance and multiple inheritance,demonstrating how to create and use them in python.

```python
## Inheritance(Single Inheritance)
## parent class

class car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
```

```python
car1=car(4,4,"petrol")
print(car1.drive())
```

```python
class tesla(car):
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype) # super().__init__(name)  >>>Python internally converts------> ParentClass.__init__(self, name)
        self.is_selfdriving=is_selfdriving
    
    def selfdriving(self):
        print(f"tesla card supports self driving:{self.is_selfdriving}")
```

```python
tesla1=tesla(4,4,"electric",True)
tesla1.selfdriving()
```

```python
tesla1.drive()
```

```python
### Multiple Inheritance 
## When a class inherits from more than one base class
# Base class 1

class Animal:
    def __init__(self,name):
        self.name= name
    
    def speak(self):
        print(f"{self.name} printed useing methos")

class Pet:
    def __init__(self,owner):
        self.owner=owner
    

## Derived class

class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    
    def speak(self):
        return f"{self.name} says woof"

## create an object 
dog=Dog("Buddy","jyothi")

```

```python
print(dog.speak())
print(dog.owner)
```

Conclusion:

Inheritance is a powerful feature in OOP that allows for code resuse and the creation of a more logical class structure.Single Inheritance invloves one base class,while multiple inheritance  involves more than on base class. Understanding how to implement and use inheritance in Python will enable you to design more efficient and maintable Object-oriented programs.
