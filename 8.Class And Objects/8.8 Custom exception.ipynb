```python
class Error(Exception):
    pass

class dobexception(Error):
    pass
```

```python
year=int(input("Enter the dob:"))
age= 2025 - year
try:

    if age<=30 and age>=20:
        print("The age is valid so you can apply for the exam")
    else:
        raise dobexception

except dobexception:
    print("age should be greater than 20 or less than 20")
```
