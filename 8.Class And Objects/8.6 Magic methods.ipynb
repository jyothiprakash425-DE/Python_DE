Magic methods in python, also known as dunder methods(double underscore methods),are spacial methods that start and end with double underscores.These methods enable you to define the behaviour of objects for built-in operations,comparisions and more.

```python
## Basic methods
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def __str__(self):
        return f"{self.name},{self.age} years old."

    def __repr__(self):
        return f"person(name={self.name},age={self.age})"

person=Person("jyothi",34)
print(person)
print(repr(person))
```
