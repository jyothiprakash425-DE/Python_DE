Polymorphism:

Ploymorphism is a core concept in Object-oriented programming(OOP) that allows objects of dirrent classes to be treated as objects of a common superclass.It provides a way to perform a single action in different forms.Polymorphism is typically achieved through methos overriding and interfaces.

Method Overriding:

Methos overriding allows a child class to provide a specific implementation of a method that is already defind in parent class.

```python
## Base class

class Animal:
    def speak(self):
        return "sound of the animal"

## Derived class 1

class Dog(Animal):
    def speak(self):
        return "woof!"

## Derived class
class cat(Animal):
    def speak(self):
        return "Meow!"

## Function that demonstrates polymorphism
def animal_speak(Animal):
    print(Animal.speak())

animal=Animal()
dog=Dog()
cat=cat()
```

```python
dog.speak()
```

```python
cat.speak()
```

```python
animal_speak(dog)
```

```python
animal_speak(animal)
```

```python
animal_speak(cat)
```

```python
### Polymorphism with functions and methods
## base class

class shape:
    def area(self):
        return "The area of the figure"

##derived class 1

class Rectangle(shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
    
    def area(self):
        return self.width*self.height

## derived class2

class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    
    def area(self):
        return 3.14*(self.radius**2)
    
# Function that demonstrates polymorphism 

def print_area(shape):
    print(f"The area is {shape.area()}")

shape=shape()
rectangle=Rectangle(4,5)
circle=circle(3)

print_area(rectangle)
print_area(circle)
print_area(shape)
```

Polymorphism with Abstract Base Classes:

Abstarct Base Classes(Abcs) are used to define common methods for a group of rlated objects.They can enforce that derived classes implement particular methods,promoting consistency across different implementations.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"


class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine started"


def start_vehicle(vehicle):
    print(vehicle.start_engine())


car_obj = Car()
bike_obj = Motorcycle()

start_vehicle(car_obj)
start_vehicle(bike_obj)
```

1.Uses abstract class
2.Rules are enforced
3.Child classes must implement abstract methods
4.Errors appear early (at object creation)
