Sets:

Sets are built-in data type in python used to store collection of unique items>they are unordered meaning that the elements do not follow a specific order,and they do not allow duplicate elements. Sets are useful for membership tests,eleiminating duplicate entries and performing mathematical set operations
like union,intersction,difference and semantic difference

```python
## create a set
my_set={1,2,3,4,5,6}
print(type(my_set))
print(my_set)
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_set=set([1,2,3,4,5,6])
print(my_set)
```

```python
my_empty_set=set([1,2,3,4,6,5,6,5])
print(my_empty_set)
```

```python
## basic set operations
## Adding and removing elements

my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
#remove the elements from a list

my_set.remove(7)
print(my_set)
```

```python
my_set.remove(10)
```

```python
my_set.discard(10)
print(my_set)
```

```python
#pop method
pop_value=my_set.pop()
print(pop_value)
print(my_set)
```

```python
## claer all the elements
my_set.clear()
print(my_set)
```

```python
## set membership test
my_set={1,2,3,4,5}
print(3 in my_empty_set)
print(10 in my_set)
```

```python
## mathematical operations

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

### union 
union_set=set1.union(set2)
print(union_set)

## intersection
intersection_set=set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
difference_set=set1.difference(set2)
print(difference_set)
```

```python
## symmetric difference
set1.symmetric_difference(set2)
```

```python
## set methods

set1={1,2,3,4,5}
set2={3,4,5}

print(set1.issubset(set2))
print(set2.issubset(set1))

print(set1.issuperset(set2))
print(set2.issuperset(set1))
```

```python
lst=[1,2,3,4,5,6]
set(lst)
```

```python
## counting unique world in text

text="jyothi prakash is father of ramtej jyothi"
words=text.split()
print(len(words))
print(words)
word_set=set(words)
print(len(word_set))
```

Conclusion:

Sets are a powerful and flexible data type in python that provide a way to store collection of unique elements.They support various operations such as Union,intersaction,difference and simmetric difference ,which are useful for mathematical computations. Understanding how to use sets and their associated methods can help you write more efficient and clean python code,especially when dealing with unique collections and membership tests
