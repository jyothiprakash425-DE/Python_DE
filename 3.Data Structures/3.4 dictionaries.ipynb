Dictionaries:

Dictionaries are unordered collections of items.they store data in key-value pairs.keys must be unique and immutable(e.g., Strings,numbers or tuples), while values can be of any type

```python
# creating dictionaries 
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
Employee_details={"name":"jyothi prakash","age":32,"Yoe":11,"location":"banglore","isactive":True}
print(Employee_details)
print(type(Employee_details))
```

```python
## accessing dictionery elements
print(Employee_details["Yoe"])
print(Employee_details["isactive"])
print(Employee_details["location"])

## Accessing elements using get() Methos

print(Employee_details.get("Yoe"))
print(Employee_details.get("isactive"))
print(Employee_details.get("location"))
```

```python
Employee_details
```

```python
## Modifying dictionaries elements 
## Dictionary are mutable,so you can add,update or delete elements

print(Employee_details)
Employee_details["age"]=35
print(Employee_details)
Employee_details["vertical"]="Ai practise"
print(Employee_details)
```

```python
#delete key and value pair

print(Employee_details)
del Employee_details["vertical"]
print(Employee_details)
```

```python
## Dictionary methods
keys=Employee_details.keys()
print(keys)
print(type(keys))

values=Employee_details.values()
print(values)
print(type(values))

items=Employee_details.items()
print(items)
print(type(items))
```

```python
##shallow copy

Employee_details_copy=Employee_details
print(Employee_details_copy)
print(Employee_details)
```

```python
Employee_details["name"]="jyothi prakash korra"
print(Employee_details)
print(Employee_details_copy)
```

```python
Employee_details_copy1["name"]="jyothi prakash"
print(Employee_details_copy1)
print(Employee_details_copy)
```

```python
## iterating over dictionaries
## you can use loops to iterate over dictionaries,keys,values or items

## Interating over keys
for keys in Employee_details.keys():
    print(keys)
```

```python
#iterating over keys

for values in Employee_details.values():
    print(values)
```

```python
#iterate over key value pairs

for key,value in Employee_details.items():
    print(f"{key}:{value}")
```

```python
#Nested Dictionaries

students={
    "student1":{"name":"jyothi","age":32},
    "student2":{"name":"prakash","age":33}
}

print(students)
```

```python
##Access nested dictinary elemnts

print(students["student1"]["name"])
print(students["student2"]["name"])
```

```python
students.items()
```

```python
#iterating over nested dictionaries:

for key,value in students.items():
    print(f"{key}:{value}")
    for item,value in value.items():
        print(f"{item}:{value}")
```

```python
## dictionary comprehension

for x in range(5):
    print(x**2)
```

```python
[x**2 for x in range(5)]
```

```python
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
#conditional dictionary comprehension

evens={x:x**2 for x in range(1,10) if x%2==0}
print(evens)
```

```python
## use a dictionary to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}
for value in numbers:
    if value in frequency:
          frequency[value]+=1
    else:
        frequency[value]=1

print(frequency)
```

```python
#merge 2 dictionaries into one

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}

merge_dict={**dict1,**dict2}
print(merge_dict)
```
