File Operation- Read And Write Files

File handling is a crucial part of any programming language.Python provides built-in functions and methods to read from and write to files,both text and binary. This lesson will cover the basics of the file handling,including read and writinh text files and binary files.

```python
# Read a whole file

with open ('persondetails.txt','r') as file:
    content=file.read()
    print(content)
    # print("test")

# Python_DE/6.File handling/
```

```python
# read a file line by line

with open('persondetails.txt','r') as file:
    # print(type(file))
    for line in file:
        print(line.strip()) ## strio() removes the newline charecter
```

```python
## writing a file(overwriting)
with open('persondetails.txt','w') as file:
    file.write("Hello Jyothi Prakash\n")
    file.write("This is a new line.")
```

```python
## write a file (without overwriting)
with open('persondetails.txt','a') as file:
    file.write("\nAppending a new line to the exisiting document.")
```

```python
## writing a list of lines to a file

lines=['First line\n','Second line\n','Third line\n']
with open('persondetails.txt','a') as file:
    for line in lines:
        file.write(line)
```

```python
## Binary files
# Writing to a binary file

data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
## reading a binary file

with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
### Read the content from source text file and write to destination text file
# copying a text file

with open('persondetails.txt','r') as source_file:
    source_content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(source_content)
```
