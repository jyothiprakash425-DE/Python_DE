Working With File paths

When working with files in python,handling file paths correctly is crucial to ensure your code works across different operating systems and environments.Python provides several modules and functions for working with the file paths effectively.

```python
#### Using the os module

import os
cwd=os.getcwd()
print(f" Current working directory is:{cwd}")
```

```python
#create a new directory

new_directory= "Bymoduleos"
os.mkdir(new_directory)
print(f"Directory :{new_directory} created")
```

```python
# listing files and dierctories

items=os.listdir('.')
print(items)
```

```python
### joining path

directory_name="jyothi"
file_name="details.txt"
fullpath= os.path.join(directory_name,file_name)
print(fullpath)
```

```python
directory_name="jyothi"
file_name="details.txt"

fullpath=os.path.join(os.getcwd(),directory_name,file_name)
print(fullpath)
```

```python
path='example1.txt'

if os.path.exists(path):
    print(f"{path} is available")
else:
    print(f"{path} doesn't exists")
```

```python
# checking if a path is a File or Directory

import os
path='persondetails.txt'

if os.path.isfile(path):
    print(f"{path} its's a file")
elif os.path.isdir(path):
    print(f"{path} it's a folder")
else:
    print(f"The path {path} is neither a file nor a directory.")

```

```python
# getting the absolute path

relative_path='persondetails.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```
