 #### Importing Modules in python : Modules and packages
In Python,modules and packages help organize and reuse code.Here's comprehensive guide on how to import them


```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
from math import *
print(sqrt(16))
print(pi)
```

```python
from package.maths import addition,substraction,multiply
print(addition(3,2))
print(substraction(3,2))
print(multiply(3,2))
```

```python
from package.subpackage.arth import mult
print(mult(3,2,4))
```

Conclusion :

Importing modules and packages in python allows you to organize your code,reuse functionalities, and keep your projects clean and manageble.By understanding how to import modules,specific functions and use relative imports within packages,you can structure your python applications more effectively
