```python
Data Types:

1. Definition:
            > Data types are a classification of data which tell the compiler or interpreter how the programmer
              intends to use the data.
            > They determine the type of operations that can be performed on the data ,the values that the 
              data can take,and the amount of memory needed to store the data.

2. Importance of Data Types in programming:
     
     Explanation:
                > Data types ensure that the data is stored in an efficient way.
                > They help in performing correct operartion on data
                > proper use of data types can prevent error and bugs in the program

3. basic data types: Integers,Floating-point number,strings and booleans
4. Advanced data Types: Lists,Tuples,Sets,Dictionaries

```

```python
#integer Example
age=35
type(age)
```

```python
# floating point data type
height =5.11
print(height)
print(type(height))
```

```python
## string data type examples

name="jyothi"
print(name)
print(type(name))
```

```python
#boolean data type
is_true=True
type(is_true)
```
