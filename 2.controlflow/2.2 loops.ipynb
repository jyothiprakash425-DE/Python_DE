```python
range(5)
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)
```

```python
#strings
str="jyothi prakash"

for i in str:
    print(i)
```

```python
str="jyothi prakash"
cnt=0

for i in str:
    if i.lower() in ("a","e","i","o","u"):
        cnt+=1
print(f"total vowels count is {cnt}")
```

```python
## While loop 

## the while loop continues to execute as long as the condition is true

count=0

while count<=5:
    print(count)
    count+=1
```

```python
## Loop control statements
## Break
## The break statement exits the loop permanently

for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue

## The continue statement skips the current iteration and continues with the next

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## Nested loops
# lop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
# calulate first N natural number using while and for loop

n=10
count=1
sum=0

while count<=10:
    sum=sum+count
    count+=1
print(f"sum of first {n} natural number :{sum}")
```

```python
n=10
sum=0
for i in range(n+1):
    sum=sum+i

print(f"sum of first {n} natural numbers is :{sum}")
```

```python
## Example- prime numbers between 1 and 100

for num in range(101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
            else:
                print(num)
```
