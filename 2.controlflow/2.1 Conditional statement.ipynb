Conditional Statements: if,elif,else

```python
## if statement

age=18

if age>=18:
    print("You are eligible to vote in the elections")
```

```python
## else
## The else statement executes a block of code if the condition in the IF Statment is false.abs

age=16

if age>=18:
    print("You are eligible to vote in the elections")
else:
    print("you are minor")
```

```python
## elif
## The elif statement allows you to check multiple conditions.It stands for "else if"

age=17

if age<13:
    print("You are child")
elif age<18:
    print("You are teenager")
else:
    print("you are an adult")

```

```python
# Nested condition statements

# you can place one or more if ,elif, or else statements inside another if,elif or else statement to create nested condition statements.

#number even,odd,negative and zero

num=int(input("Enter Number:"))

if num>0:
    print(f"User Entered number {num} is positive")
    if num%2==0:
        print(f"user entered number is {num} is even")
    else:
        print(f"user eneterd number is {num} is odd")
else:
    print("Entered number is Zero or Negative")


  
```

```python
## determine year is leap year or not

year_val=int(input("Enter A year number"))

if year_val%4==0:
    if year_val%100==0:
        if year_val%400==0:
            print(f"{year_val} is aleap year")
        else:
            print(f"{year_val} is not a leap year")
    else:
         print(f"{year_val} is a leap year")
else:
    print(f"{year_val} is not a leap year")
```

```python
## simple calculator program

num1=int(input("Enter first number:"))
num2=int(input("Enter second number:"))
Operator=input("Enter the operator type(+,-,*,/):")

#perform the requested operation

if Operator=='+':
    result=num1+num2
elif Operator=='-':
    result=num1-num2
elif Operator=='*':
    result=num1*num2
elif Operator=='/':
    if num2!=0:
        result= num1/num2
    else:
        result="Error! Divisible by Zero"
else:
    result="Invalid Operation"

print("result:",result)
```
