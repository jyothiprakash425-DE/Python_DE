```python
#Example 1.Manage To do list

to_do_list=["buy groceries","clean the house","pay bills"]

##adding to task
to_do_list.append("schedule meeting")
to_do_list.append("go for a run")

#removing the task

to_do_list.remove("clean the house")

#cl=hecking if a task in a list

if "pay bills" in to_do_list:
    print("don't forget to pay the bills")

print("To do list remaining")

for task in to_do_list:
    print(f"-{task}")
```

```python
# create a list and calculate average grades for students

grades=[85.92,78,90,88]

#adding a new grade
grades.append(99)

#calculate the average grade

average_grade = sum(grades) / len(grades)

# finding max and min grades
highest_grade=max(grades)
min_grades=min(grades)

print(f"Average grade value:{average_grade:.2f}")
print(f"minimum grade:{min_grades}")
print(f"maximum grade:{highest_grade}")
```

```python
# managing an inventory

inventory=["apples","banana","oranges","grapes"]

#adding a new item
inventory.append("strawberry")

# removing an item this is out of stock
inventory.remove("oranges")

item_name="oranges"

if item_name in inventory:
    print(f"{item_name} are available")
else:
    print(f"{item_name} are out of stock")

#printinh te inventory

for item in inventory:
    print(f"-{item}")
```

```python
# use a list and analyze user feedback

feedback=["Great Service!","very staisfied","could be better","Excellent experience"]

#Adding new feedback
feedback.append("not happy with service")

#counting specific feedback

positive_feedback_count=sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())

print(f"positive feedback count:{positive_feedback_count}")

#printing all feedback

for comment in feedback:
    print(f"-{comment}")

```
