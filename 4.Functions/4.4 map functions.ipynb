The map() function in Python

The map() function applies a given function to all items in an inout list(or any other iterable) and returns a map object (an iterator).This is particularly useful for transforming data in a list comprehensively.

```python
def square(x):
    return x**2

square(4)
```

```python
numbers=[1,2,3,4,5,6,7,8]
list(map(square,numbers))
```

```python
## lambda function with map
numbers=[1,2,3,4,5,6,7,8]
list(map(lambda x:x**2,numbers))
```

```python
#### Map multiple iterables
num1=[1,2,3]
num2=[4,5,6]

add_numbers=list(map(lambda x,y:x+y,num1,num2))
print(add_numbers)
```

```python
## map() to convert a list of strings to integers
# use map to convert string to integer
in_string=['1','2','3','4']

result=list(map(int,in_string))
print(result)
```

```python
fruits=["appLe","Banana","grapes"]

result=list(map(str.upper,fruits))
print(result)
```

```python
def find_names(details):
    return details["name"]


people=[
    {"name":"jyothi","age":34},
    {"name":"prakash","age":35}
]

list(map(find_names,people))
```
